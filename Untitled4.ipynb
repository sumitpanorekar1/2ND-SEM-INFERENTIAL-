```python
import scipy.stats as st
O = [8, 6, 10, 7, 8, 11, 9]
E = [9, 8, 11, 8, 10, 7, 6]
chi = 0

for i in range(len(O)):
    chi = chi + ((O[i] - E[i]) ** 2) / E[i]
print("Chi Square Value =", chi)

df = len(O) - 1

critical = st.chi2.ppf(0.95, df)
print("Critical Value =", critical)

if chi < critical:
    print("Accept H0")
    print("Goodness of fit exists")
else:
    print("Reject H0")
```

```python



from math import comb
import matplotlib.pyplot as plt

n = 6          
p = 0.5       
k = 2          

probability = comb(n, k) * (p ** k) * ((1 - p) ** (n - k))

print("Probability of getting exactly 2 heads =", probability)
x = []
y = []

for i in range(n + 1):
    prob = comb(n, i) * (p ** i) * ((1 - p) ** (n - i))
    x.append(i)
    y.append(prob)
print("\nBinomial Distribution:")
for i in range(len(x)):
    print(f"Heads = {x[i]}, Probability = {y[i]:.6f}")

# Plot the distribution
plt.figure(figsize=(8,5))
plt.bar(x, y)

plt.xlabel("Number of Heads")
plt.ylabel("Probability")
plt.title("Binomial Distribution for 6 Coin Tosses")
plt.xticks(x)

plt.show()
```
